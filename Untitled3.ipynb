```python
# a. Menyimpan nilai tegangan dan arus, lalu menghitung daya listrik
tegangan = 220  # Volt
arus = 15       # Ampere
daya = tegangan * arus
print(f"Daya listrik yang dihasilkan generator: {daya} Watt")

# b. Menghitung efisiensi turbin
energi_input = 500
energi_output = 400
efisiensi = (energi_output / energi_input) * 100
print(f"Efisiensi turbin: {efisiensi:.2f}%")
```

```python
# Data suhu pada 5 titik boiler (°C)
suhu = [92.5, 88.7, 95.1, 102.3, 97.8]

# a. Tampilkan suhu pada titik ke-3
print(f"Suhu pada titik ke-3: {suhu[2]} °C")

# b. Tampilkan tiga pembacaan pertama menggunakan slicing
tiga_pertama = suhu[:3]
print(f"Tiga pembacaan pertama: {tiga_pertama}")

# c. Tambahkan pembacaan baru 99.4 ke dalam list, lalu tampilkan jumlah data
suhu.append(99.4)
jumlah_data = len(suhu)
print(f"List suhu setelah penambahan: {suhu}")
print(f"Jumlah data setelah penambahan: {jumlah_data}")
```

```python
def tentukan_status_katup(tekanan_uap, status_manual_override):
    if status_manual_override == True:
        return "Katup Mode Manual"
    elif tekanan_uap > 150 and status_manual_override == False:
        return "Katup Terbuka"
    else:
        return "Katup Tertutup"

# Uji program dengan kombinasi yang diminta
uji_kasus = [
    (160, False),
    (160, True),
    (140, False),
    (150, False),
    (140, True)
]

print(f"{'Tekanan':<10} | {'Override':<10} | {'Output yang Diharapkan':<22} | {'Output Program'}")
print("-" * 70)

for tek, ovr in uji_kasus:
    status = tentukan_status_katup(float(tek), ovr)
    print(f"{tek:<10} | {str(ovr):<10} | {status:<22} | {status}")
```

```python
suhu_saat_ini = 60
target_suhu = 85

while suhu_saat_ini < target_suhu:
    suhu_saat_ini += 5
    print(f"Suhu saat ini: {suhu_saat_ini} °C")

    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

print("Target suhu telah tercapai!")
```

```python
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

data_tidak_valid = 0

for i, nilai_str in enumerate(data, start=1):
    try:
        nilai = float(nilai_str)

        if nilai > 2.0:
            print(f"Warning: Unit [{i}] vibrasi tinggi!")
            if nilai > 3.0:
                print("KRITIS: pemeriksaan dihentikan")
                break

    except ValueError:
        print(f"Unit [{i}]: data tidak valid")
        data_tidak_valid += 1
        continue

print(f"Jumlah data tidak valid: {data_tidak_valid}")
```
